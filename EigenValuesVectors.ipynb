```python
#Data
mat = [
  [9.07, -9.21, -0.73],
  [-9.21, 129.93, 91.54],
  [-0.73, 91.54, 684.7]
]
```

```python
import numpy as np
mat=np.array(mat)
eigen_values, eigen_vectors = np.linalg.eig(mat)
print("EigenValues:", np.round(eigen_values, 3))
print("eigenVectors:")
# print(eigen_vectors)
print(np.round(eigen_vectors, 3))
```
