```python
#inverse matrix

import numpy as np
mat = [
  [9.07, -9.21, -0.73],
  [-9.21, 129.93, 91.54],
  [-0.73, 91.54, 684.7]
]
mat= np.array(mat)
inverse=np.linalg.inv(mat)
# print(inverse)
print(np.round(inverse, 5))
```
