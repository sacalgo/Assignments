```python
#Data
H=[1.70, 1.62, 1.52, 1.85, 1.91, 1.42]
W=[72, 64, 84, 80, 72, 70]
n=len(H)
```

```python
#Mean
mean_H = sum(H)/n
mean_W = sum(W)/n
print("Mean H=", mean_H)
print("Mean W=", mean_W)
```

```python
#Deviations
x=[]
for h in H:
  x.append(h-mean_H)
y=[]
for w in W:
  y.append(w-mean_W)
```

```python
#Required sums
sum_x2,sum_y2, sum_xy=0, 0, 0
for a in x:
  sum_x2 += a*a
for b in y:
  sum_y2+= b*b
for a, b in zip(x, y):
  sum_xy += a*b

```

```python
#covariance matrix
cov_HH = sum_x2/(n-1)
cov_HW = sum_xy/(n-1)
cov_WW = sum_y2/(n-1)
```

```python
covarince_matrix = [
    [cov_HH, cov_HW],
    [cov_HW, cov_WW]
]
print("\nCovariance Matrix:")
for row in covarince_matrix:
   for value in row:
     print(round(value, 5), end=" ")
   print()
```
