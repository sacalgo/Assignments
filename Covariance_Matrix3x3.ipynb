```python

#DATA
x=[12.1, 13.2, 15.6, 17.2, 18.8, 10.3, 11.7, 16.4]
y=[48, 54, 32, 18, 41, 32, 31, 30]
z=[101, 171, 112, 132, 140, 112, 151, 96]
n=len(x)
```

```python
#mean
mean_X= round(sum(x)/n, 2)
mean_Y= round(sum(y)/n, 2)
mean_Z=round(sum(z)/n, 2)
print(mean_X, mean_Y, mean_Z)
```

```python
#deviations
x1=[]
for i in x:
  x1.append(round(i-mean_X, 2))
y1=[]
for i in y:
  y1.append(round(i-mean_Y, 2))
z1=[]
for i in z:
  z1.append(round(i-mean_Z, 2))

print(x1)
print(y1)
print(z1)
```

```python
#required sums
sum_xx, sum_yy, sum_zz=0, 0, 0

for i in x1:
  sum_xx += i*i
for i in y1:
  sum_yy += i*i
for i in z1:
  sum_zz += i*i

sum_xy, sum_yz, sum_xz=0, 0, 0
for i, j in zip(x1, y1):
  sum_xy += i*j
for i, j in zip(x1, z1):
  sum_xz += i*j
for i, j in zip(y1, z1):
  sum_yz += i*j

print(sum_xx, sum_yy, sum_zz, sum_xy, sum_xz, sum_yz)
```

```python
#covariance matrix values
cov_xx= sum_xx/(n-1)
cov_yy= sum_yy/(n-1)
cov_zz=sum_zz/(n-1)
cov_xz=sum_xz/(n-1)
cov_xy=sum_xy/(n-1)
cov_yz=sum_yz/(n-1)
```

```python
covariance_matrix=[
    [cov_xx, cov_xy, cov_xz],
    [cov_xy, cov_yy, cov_yz],
    [cov_xz, cov_yz, cov_zz]
]
print("\nCovariance Matrix:")
for row in covariance_matrix:
  for value in row:
    print(round(value, 3), end=" ")
  print()
```
